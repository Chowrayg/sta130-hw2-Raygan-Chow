# Research question 1

Which groups of people do people have the most value-to-time-spent?
The question aims to draw out a statistical support or rejection of what group of people we should prioritize spending time with.


## Variables 

CONNECTION_social_time_family_p7d_grouped describes the question: In the PAST WEEK, how many hours in total did you spend socializing with others from family members?

CONNECTION_social_time_friends_p7d_grouped describes the question: In the PAST WEEK, how many hours in total did you spend socializing with others from friends?

CONNECTION_social_time_coworkers_and_classmates_p7d_grouped describes the question: In the PAST WEEK, how many hours in total did you spend socializing with others from Coworkers or Classmates?

CONNECTION_social_time_neighbours_p7d_grouped describes the question: In the PAST WEEK, how many hours in total did you spend socializing with others from neighbours?

CONNECTION_meaningful_time_family_grouped describes the question:	Generally speaking, how much of your time spent with family members is meaningful and fulfilling?

CONNECTION_meaningful_time_friends_grouped describes the question: 	Generally speaking, how much of your time spent with friends is meaningful and fulfilling?

CONNECTION_meaningful_time_coworkers_grouped describes the question: 	Generally speaking, how much of your time spent with coworkers or classmates is meaningful and fulfilling?

CONNECTION_meaningful_time_neighbours_grouped describes the question: 	Generally speaking, how much of your time spent with neighbours is meaningful and fulfilling?

The first 4 variables are used to visualize the distribution of hours per week spent for each group, and ultimately the denominator of value-to-time using their means.

The latter 4 variables are used to visualize the distribution of fulfilling time per week spent for each group, and ultimately the numerator of value-to-time using their means.

Each variable's data is first represented using a histogram as a histogram is useful for categorizing hours per week spent. The value-to-time result for each group of people are represented in a single bar plot as bar plots are suitable for non-numeric categories such as groups in this case. Value to time of each group of people are pitted against each other within this bar plot, calculation and reasoning to be discussed in the analysis section.

## Analysis

The analysis consists of 4 steps.
1. histograms of each variable's data are created using px.histogram with suitable nbins
2. The mean of all 8 variables respectively are calculated using the aggregation function .mean()
3. Divide the first 4 variables' means by their matching latter 4 variables, forming the value to time ratio
4. bar plot containing value to time ratio of the 4 people groups to see which group is most worth spending time

We assume that the mean for each variable serve as reliable numerators and denominators for the ratio regardless of skewing; that the answering speed is increased since it's likely participants had done a cross-sectional survey before while maintaining answer accuracy; that the participants who answered the first 4 questions also answered the respective latter 4 questions so the ratio uses the data from the same samples.

## Hypothesis

Null: Family has the highest value to time ratio.
Alternative: Family does not have the highest value to time ratio.

If there isn't sufficient evidence to reject the null, it answers the research question directly. If there is, it potentially reflects a social phenomenon where it is more worthwhile for the younger generations to spend time with say friends than with parents, signifying either a switch in parents' roles or a significant problem of wavering ties.

# Research question 2
Is there a relationship between quality of life and time spent with coworkers or classmates?

## Variables 
WELLNESS_life_satisfaction describes the question: On a scale of 1 to 10, How do you feel about your life as a whole right now?

CONNECTION_social_time_coworkers_and_classmates_p7d_grouped describes the question: In the PAST WEEK, how many hours in total did you spend socializing with others from the following groups? - Coworkers or Classmates

The variables aren't time based in spite of question 2's description but grouped by x to y hours, so it is difficult to implement KDE. For WELLNESS_life_satisfaction, the scale is 1 to 10 which is not that many, thus a bar plot is enough to explain both.

## Analysis

Using the .describe() method, we can further explore the distribution of life satisfaction scores within each social time group. This provides insights into the spread and central tendency of life satisfaction within each time category.

If there are missing values in the CONNECTION_social_time_coworkers_and_classmates_p7d_grouped or WELLNESS_life_satisfaction columns, pandas will handle them appropriately during aggregation. The .describe() method will automatically exclude missing values when calculating summary statistics
## Hypothesis
Null: max of WELLNESS_life_satisfaction is 7 and max of CONNECTION_social_time_coworkers_and_classmates_p7d_grouped is less than 1 hour
Alternative: max of WELLNESS_life_satisfaction is not 7 and max of CONNECTION_social_time_coworkers_and_classmates_p7d_grouped is not less than 1 hour

If there isn't sufficient evidence to reject the null, it shows that social time with coworkers and classmates isn't necessary to life satisfaction. If there is, depending on which statistic is inaccurate, social time with coworkers and classmates may affect life satisfaction positively when WELLNESS_life_satisfaction is low or when max of CONNECTION_social_time_coworkers_and_classmates_p7d_grouped is a lot of hours.
